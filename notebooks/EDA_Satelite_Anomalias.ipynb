# Análisis de Anomalías Satelitales (VIIRS / Sentinel-2)
Este notebook explora las luces nocturnas y busca anomalías extremas como el incendio de agosto de 2023.

```
import rasterio
import matplotlib.pyplot as plt
import numpy as np
import os
```

```
def plot_viirs(filepath, title):
    with rasterio.open(filepath) as src:
        img = src.read(1)
        # Enmascarar valores negativos (nodata)
        img = np.where(img < 0, np.nan, img)
        max_val = np.nanmax(img)
        
        plt.figure(figsize=(10,8))
        plt.imshow(img, cmap='magma', vmin=0, vmax=200)
        plt.colorbar(label='Radianza (nW/cm²/sr)')
        plt.title(f'{title}\nValor Máximo Detectado: {max_val:.1f}')
        plt.axis('off')
        plt.show()
```

```
# Comparar un mes normal (Julio 2023) con el mes del incendio (Agosto 2023)
viirs_jul = '../data/Satelite_VIIRS/tenerife_viirs_2023_07.tif'
viirs_ago = '../data/Satelite_VIIRS/tenerife_viirs_2023_08.tif'

if os.path.exists(viirs_jul) and os.path.exists(viirs_ago):
    plot_viirs(viirs_jul, 'VIIRS - Julio 2023 (Normal)')
    plot_viirs(viirs_ago, 'VIIRS - Agosto 2023 (Incendio Forestal)')
else:
    print('Archivos no encontrados')
```
