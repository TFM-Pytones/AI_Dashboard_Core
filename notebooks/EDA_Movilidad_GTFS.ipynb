# EDA - Movilidad (TITSA) vs Zonas Turísticas
Superponer paradas de guaguas con polígonos de zonas turísticas para ver dónde falta transporte.

```
import geopandas as gpd
import matplotlib.pyplot as plt
from sqlalchemy import create_engine
import os
from dotenv import load_dotenv
```

```
load_dotenv('../.env')
engine = create_engine(f"postgresql://{os.getenv('AZURE_DB_USER')}:{os.getenv('AZURE_DB_PASSWORD')}@{os.getenv('AZURE_DB_HOST')}:5432/{os.getenv('AZURE_DB_NAME')}?sslmode=require")
```

```
# Intentar cargar Paradas y Zonas Turísticas desde PostgreSQL (esquema bronze)
try:
    paradas = gpd.read_postgis('SELECT * FROM bronze.gtfs_paradas', engine, geom_col='geometry')
    zonas = gpd.read_postgis('SELECT * FROM bronze.zonas_turisticas', engine, geom_col='geometry')
    
    # Visualización
    fig, ax = plt.subplots(figsize=(12, 12))
    zonas.plot(ax=ax, color='lightblue', edgecolor='blue', alpha=0.5, label='Zonas Turísticas')
    paradas.plot(ax=ax, color='red', markersize=2, alpha=0.3, label='Paradas TITSA')
    
    plt.title('Densidad de Paradas de Guaguas vs Zonas Turísticas')
    plt.legend()
    plt.show()
except Exception as e:
    print("Aún no tienes estas tablas en la base de datos de Azure. Súbelas primero con Python.")
```
