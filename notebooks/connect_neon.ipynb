```python
import os
import psycopg2
from dotenv import load_dotenv

# Forzamos la carga del nuevo .env
load_dotenv(override=True)

try:
    print("Iniciando conexión limpia...")
    conn = psycopg2.connect(
        user=os.getenv("DB_USER"),
        password=os.getenv("DB_PASSWORD"),
        host=os.getenv("DB_HOST"),
        database=os.getenv("DB_NAME"),
        port="5432",
        sslmode="require"
    )
    print("Conexión completada")
    
    # Comprobamos que podemos hablar con la base de datos
    cursor = conn.cursor()
    cursor.execute("SELECT postgis_full_version();")
    print(f"PostGIS listo: {cursor.fetchone()[0][:45]}...")
    
    cursor.close()
    conn.close()

except Exception as e:
    print(f"Algo ha fallado: {e}")
```
