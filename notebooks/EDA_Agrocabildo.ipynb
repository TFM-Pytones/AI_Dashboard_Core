# Análisis Exploratorio de Datos (EDA) - Agrocabildo
Este notebook busca anomalías en los datos climáticos.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

plt.style.use('ggplot')
```

```python
# Cargar datos de clima y estaciones
df_clima = pd.read_parquet('../data/clima_horario_agrocabildo.parquet')
df_estaciones = pd.read_parquet('../data/estaciones_agrocabildo.parquet')

print(f'Registros clima: {len(df_clima):,}')
```

```python
# 1. Revisar valores nulos
print(df_clima.isnull().sum())
```

```python
# 2. Distribución de Temperaturas para detectar anomalías
plt.figure(figsize=(10,5))
sns.histplot(df_clima['temperatura'].dropna(), bins=100)
plt.title('Distribución de Temperaturas (Tenerife)')
plt.show()
```

```python
# 3. Días con temperaturas extremas (Ola de calor / Incendios)
extremos = df_clima[df_clima['temperatura'] > 40]
display(extremos.sort_values('temperatura', ascending=False).head(15))
```
