```python
import os
import psycopg2
from dotenv import load_dotenv

# Carga automáticamente las variables guardadas en el archivo .env
load_dotenv()

# Accede a la variable de entorno de forma segura
DATABASE_URL = os.getenv("DATABASE_URL")

def conectar_base_datos():
    try:
        # Establece la conexión con Neon
        conn = psycopg2.connect(DATABASE_URL)
        print("🚀 Conexión segura establecida con Neon.tech")
        return conn
    except Exception as e:
        print(f"❌ Error crítico al conectar a la base de datos: {e}")
        return None

if __name__ == "__main__":
    conexion = conectar_base_datos()
    if conexion:
        conexion.close()
```
